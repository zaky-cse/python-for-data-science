```python
students = {"Rahim": [80, 75, 90], "Karim": [60, 55, 70], "Salma": [95, 88, 92]}

def average(numbers):
    return sum(numbers) / len(numbers)

for name, marks in students.items():
    avg = average(marks)
    if avg >= 80:
        grade = "A"
    elif avg >= 60:
        grade = "B"
    else:
        grade = "C"
    print(name, "-", round(avg, 1), "-", grade)
```
